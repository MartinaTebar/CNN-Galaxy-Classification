```python
import h5py
import numpy as np


with h5py.File('../data/Galaxy10_DECals_NoDuplicated.h5', 'r') as F:
    images = np.array(F['images'])   
    labels = np.array(F['ans']) 
         

print(images.shape, images.dtype)
print(labels.shape, labels.dtype)
```

```python
from sklearn.model_selection import train_test_split


train_idx, test_idx = train_test_split(
    np.arange(labels.shape[0]),
    test_size=0.1,
    stratify=labels,
    random_state=42
)

print(f'Train: {len(train_idx)}, Test: {len(test_idx)}')


# train_images, train_labels, test_images, test_labels = images[train_idx], labels[train_idx], images[test_idx], labels[test_idx]
```

```python
train_idx2, val_idx = train_test_split(
    train_idx,
    test_size=0.1111, 
    stratify=labels[train_idx],
    random_state=42
)

train_images, train_labels = images[train_idx2], labels[train_idx2]
val_images, val_labels     = images[val_idx],    labels[val_idx]
test_images, test_labels   = images[test_idx],   labels[test_idx]

print(train_images.shape, val_images.shape, test_images.shape)
```
